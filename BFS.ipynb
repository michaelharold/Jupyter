```python
from collections import deque

def bfs(graph, start):
    """
    Perform Breadth-First Search (BFS) on an undirected or directed graph.
    
    Parameters:
        graph (dict): adjacency list representation of the graph
        start: starting node

    Returns:
        list: BFS traversal order
    """
    visited = set()
    queue = deque([start])
    traversal = []

    visited.add(start)

    while queue:
        node = queue.popleft()
        traversal.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return traversal



graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "F"],
    "D": ["B"],
    "E": ["B", "F"],
    "F": ["C", "E"]
}

result = bfs(graph, "A")
print("BFS Traversal:", result)
```
