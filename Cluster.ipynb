```python
import numpy as np

def kmeans_euclidean(X, n_clusters, max_iter=300, tol=1e-4, random_state=None):
    """Cluster rows of X using K-means and Euclidean distance."""
    X = np.asarray(X, dtype=float)

    if X.ndim != 2 or len(X) == 0:
        raise ValueError("X must be a non-empty 2D array.")
    if not 1 <= n_clusters <= len(X):
        raise ValueError("n_clusters must be between 1 and the number of samples.")

    rng = np.random.default_rng(random_state)
    centers = X[rng.choice(len(X), size=n_clusters, replace=False)].copy()

    for _ in range(max_iter):
        # Euclidean distance from each sample to each center
        distances = np.linalg.norm(X[:, None, :] - centers[None, :, :], axis=2)
        labels = np.argmin(distances, axis=1)

        new_centers = centers.copy()
        for cluster in range(n_clusters):
            members = X[labels == cluster]
            if len(members):
                new_centers[cluster] = members.mean(axis=0)

        if np.linalg.norm(new_centers - centers) <= tol:
            centers = new_centers
            break
        centers = new_centers

    distances = np.linalg.norm(X[:, None, :] - centers[None, :, :], axis=2)
    labels = np.argmin(distances, axis=1)
    return labels, centers

```

```python
# Sample data points
X = np.array([
    [1.0, 2.0],
    [1.5, 1.8],
    [5.0, 8.0],
    [8.0, 8.0],
    [1.0, 0.6],
    [9.0, 11.0]
])


labels, centers = kmeans_euclidean(X, n_clusters=2, random_state=0)


print("Cluster labels:", labels)
print("Cluster centers:")
print(centers)
```
